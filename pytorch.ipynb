```python
import torch
import torch.nn as nn

model=nn.Sequential(
    nn.Linear(10,5),
    nn.ReLU(),
    nn.Linear(5,2)
)

x=torch.randn(4,10) #data 4 rows and 10 columns
print(x)
output=model(x)
print(output)
print(output.shape)
```

```python
loss_fn = nn.CrossEntropyLoss()
y_true =torch.tensor([0,1,1,0])
loss=loss_fn(output,y_true)
print(loss)
```

```python
optimizer=torch.optim.Adam( #adam optimizer we are using here
    model.parameters(), #will bring weights and bais from model
    lr=0.001 #learning rate
)
optimizer.zero_grad() #stored gradient ko change karta hai new calculation ke liye, but 0 nhi karta remember
loss.backward() #backpropagation
optimizer.step() #update weights and baisede with calculated value of gradient decent and lr so that model in next step can do good prediction
```

```python
new_output=model(x)
print(new_output) #after backpropogation shifted values of weights and baised new output we got
print(new_output.w)
```

```python
for name,param in model.named_parameters(): #isse hmm sare updated layers ke weights and baises check kar sakte hai
    print(name)
    print(param)
```

DataSet +  Dataloader

```python
from torch.utils.data import TensorDataset
from torch.utils.data import DataLoader

x = torch.randn(100,10) #input 100 rowns and 10 columns
y = torch.randint(0,2,(100,)) #output 1oo rows value will be either 0 or 1

dataset= TensorDataset(x,y) #will map x with y

loader=DataLoader(
    dataset,
    batch_size=16,#in one batch 16 rows will go
    shuffle=True
)

for batch_x,batch_y in loader:
    print(batch_x[0])
```

```python
#Project Training Loop
for epoch in range(5):
    for batch_x,batch_y in loader:
        optimizer.zero_grad()
        output=model(batch_x)
        loss=loss_fn(output,batch_y)
        loss.backward()
        optimizer.step()
    print(epoch,loss.item())
```

```python
print(output)
print("Actual labels:", batch_y[:5])
print("Predictions:", output.argmax(dim=1)[:5]) #dim=1 means har row ke andar column ko compare karo and dim=0 for vice versa
print("Raw output:", output[:5])
```
